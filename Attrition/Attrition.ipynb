```python
# Converting into table
df = pd.DataFrame(data)
df
```

```python
# Dropping identifier column
new_df = df.drop(columns=['EmployeeID'])
new_df
```

```python
# Encoding categorical variables

# Binary Encoding
new_df.replace({
    "Yes":1,
    "No":0
}, inplace=True)

# Ordinal Encoding
new_df["WorkLifeBalance"] = new_df["WorkLifeBalance"].replace({
    "Poor":0,
    "Average":1,
    "Good":2,
    "Excellent":3
})

# One-Hot Encoding
new_df = pd.get_dummies(
    new_df,
    columns=[
        "BusinessTravel",
        "JobRole"
    ],
    drop_first=True,
    dtype=int
)
```

```python
new_df
```

```python
X = new_df.drop("Resigned", axis=1).values

y = new_df["Resigned"].values
```

```python
# Normalize the features
mean = X.mean()
std = X.std()
X_scale = (X-mean)/std
```

```python
# Initialize before training

m,n = X_scale.shape  # m -> number of samples, n -> number of features
weights = np.zeros(n)
bias = 0

learning_rate = 0.01
epochs = 3000
```

```python
# Sigmoid function

def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
# Logistic Regression Algorithm

for epoch in range(epochs):
    # Forward pass
    z = np.dot(X_scale,weights) + bias
    prediction = sigmoid(z)

    # Loss Function
    loss = -np.mean(
        y * np.log(prediction + 1e-15)
        +
        (1-y) *
        np.log(
            1-prediction + 1e-15
        )
    )

    # Gradients
    dw = (1/m) * np.dot(X_scale.T, (prediction-y))
    db = (1/m) * np.sum(prediction-y)

    # Update
    weights -= learning_rate*dw
    bias -= learning_rate*db

    # Print Loss
    if epoch % 500 == 0:
        print(
            f"Epoch {epoch}, Loss = {loss:.4f}"
        )

print("\nTraining Completed")

print("\nWeights", weights)
print("\nBias", bias)
```
